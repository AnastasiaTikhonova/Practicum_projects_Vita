# Вита

Дипломный проект: формирование бэклога и моделирование процессов для сети клиник «Вита».

Тетрадь фиксирует результаты работы: интервью с заинтересованными лицами, карту пользовательских историй, границу MVP, прототип интерфейса в Figma и модели потоков данных (контекстная и логическая DFD).


## Описание проекта

К компании RANA.Software обратилась сеть медицинских клиник **«Вита»**. Заказчик хочет развивать приложение для пациентов. Часть продукта уже работает: пациент может записаться на приём и посмотреть свои записи, врач — вести карты пациентов. Не закрыт следующий контур: пациент должен видеть свои назначения и результаты анализов и понимать, в какое время анализы можно сдать.

Цель проекта — не писать серверный код, а собрать требования к доработке, ограничить первую поставку и показать, как новые функции встраиваются в процессы клиники. Результат работы — бэклог с границей MVP, прототип интерфейса и модели потоков данных.

Работа шла в несколько шагов.

1. Провести интервью с заинтересованными лицами и выделить болевые точки.
2. Разработать карту пользовательских историй.
3. Определить MVP первой доработки.
4. Собрать прототип интерфейса в Figma.
5. Смоделировать процессы на контекстной и логической DFD.

## Шаг 4. Контекстная и логическая DFD

Диаграммы хранятся в файле `Working_file_DFD.drawio`.

**Контекстная DFD.** Система — приложение «Вита». Внешние сущности: пациент, врач, лаборатория. Пациент передаёт данные записи и получает подтверждение, направления и результаты. Врач передаёт направления на обследование, сдачу анализов и медикаментозное лечение и получает данные о пациенте. Лаборатория принимает заказ и возвращает результаты анализов.

**Логическая DFD.** Процесс 0 декомпозирован на два процесса первого уровня:

- **1 — Пройти обследование:** направление, заказ в лабораторию, результаты анализов;
- **2 — Пройти медикаментозное лечение:** просмотр назначения, скачивание PDF, запрос и получение статуса выполнения.

Хранилища: D1 «Пациенты», D2 «Результаты анализов», D3 «Назначения».

Ячейка ниже читает файл диаграммы и выводит страницы, сущности, процессы, хранилища и подписанные потоки.


```python
import html
import re
import xml.etree.ElementTree as ET

drawio = ROOT / 'Working_file_DFD.drawio'
tree_root = ET.parse(drawio).getroot()


def cell_text(raw):
    text = html.unescape(raw or '')
    text = re.sub(r'<br\s*/?>', ' ', text)
    text = re.sub(r'<[^>]+>', ' ', text)
    text = text.replace('\xa0', ' ').replace('&nbsp;', ' ')
    return re.sub(r'\s+', ' ', text).strip()


SKIP_MARKERS = (
    'задание',
    'заметка для ревьюера',
    'элементы для построения',
    'текст заметки',
    'как выполнить',
)


def keep(text):
    low = text.lower()
    return text and not any(marker in low for marker in SKIP_MARKERS)


print('Страницы файла Working_file_DFD.drawio:')
for i, diagram in enumerate(tree_root.findall('diagram'), start=1):
    print(f'  {i}. {diagram.get("name")}')

context = next(d for d in tree_root.findall('diagram') if 'контекст' in (d.get('name') or '').lower())
logical = next(d for d in tree_root.findall('diagram') if 'логич' in (d.get('name') or '').lower())


def collect(diagram):
    entities, processes, stores, flows = [], [], [], []
    seen = set()
    for cell in diagram.iter('mxCell'):
        style = cell.get('style') or ''
        value = cell_text(cell.get('value'))
        if not keep(value):
            continue
        key = value.lower()
        if key in seen:
            continue
        if 'dataStoreID' in style or value.upper().startswith('D'):
            stores.append(value)
            seen.add(key)
        elif 'swimlane' in style or value in {'0', '1', '2'}:
            continue
        elif any(name in value for name in ('Приложение', 'Пройти')):
            processes.append(value)
            seen.add(key)
        elif value in {'Пациент', 'Врач', 'Лаборатория'}:
            entities.append(value)
            seen.add(key)
        elif cell.get('edge') == '1' or 'edgeLabel' in style or (
            'text;' in style and 'strokeColor=none' in style
        ):
            if 'mxgraphmodel' in key or value.startswith('%3C'):
                continue
            flows.append(value)
            seen.add(key)
    return entities, processes, stores, flows


print()
print('Контекстная DFD')
entities, processes, stores, flows = collect(context)
print('  Внешние сущности:')
for item in entities:
    print(f'    - {item}')
print('  Процесс 0:')
for item in processes:
    print(f'    - {item}')
print('  Потоки данных:')
for item in flows:
    print(f'    - {item}')

print()
print('Логическая DFD')
entities, processes, stores, flows = collect(logical)
print('  Внешние сущности:')
for item in entities:
    print(f'    - {item}')
print('  Процессы 1 уровня:')
for item in processes:
    print(f'    - {item}')
print('  Хранилища:')
for item in stores:
    print(f'    - {item}')
print('  Потоки данных:')
for item in flows:
    print(f'    - {item}')

assert 'Пациент' in entities
assert any('обследование' in item.lower() for item in processes)
assert any('лечение' in item.lower() for item in processes)
assert any(item.upper().startswith('D1') for item in stores)
print()
print('Проверка модели: внешние сущности, два процесса 1 уровня и хранилища D1–D3 на месте.')
```

Ниже воспроизведён сквозной сценарий пациента по логической DFD: направление на обследование, заказ в лабораторию, результат в хранилище D2, назначение на лечение в D3, просмотр, PDF и статус.


```python
from datetime import date

lab_hours = {
    'пн-пт': '07:30–12:00',
    'сб': '08:00–11:00',
    'вс': 'выходной',
}

patient = {
    'id': 'P-1042',
    'name': 'Пациент из интервью',
    'store': 'D1 Пациенты',
}

exam = {
    'type': 'Направление на обследование',
    'process': '1. Пройти обследование',
    'tests': ['ОАК', 'глюкоза', 'АЛТ'],
    'order_to': 'Лаборатория',
}

lab_result = {
    'store': 'D2 Результаты анализов',
    'status': 'готов',
    'date': date(2026, 3, 12).isoformat(),
    'values': {'ОАК': 'в норме', 'глюкоза': '5.2 ммоль/л', 'АЛТ': 'в норме'},
}

treatment = {
    'type': 'Назначение на медикаментозное лечение',
    'process': '2. Пройти медикаментозное лечение',
    'store': 'D3 Назначения',
    'status': 'выполняется',
    'pdf': 'naznachenie_P-1042.pdf',
}

print('Сквозной сценарий MVP')
print(f"  Пациент: {patient['id']}  ({patient['store']})")
print(f"  Процесс 1: {exam['process']}")
print(f"    направление: {exam['type']}")
print(f"    анализы: {', '.join(exam['tests'])}")
print(f"    заказ: {exam['order_to']}")
print('    время сдачи:')
for day, hours in lab_hours.items():
    print(f'      {day}: {hours}')
print(f"  Результат: {lab_result['store']}, статус «{lab_result['status']}», дата {lab_result['date']}")
for name, value in lab_result['values'].items():
    print(f'      {name}: {value}')
print(f"  Процесс 2: {treatment['process']}")
print(f"    назначение: {treatment['type']}")
print(f"    хранилище: {treatment['store']}")
print(f"    статус: {treatment['status']}")
print(f"    файл: {treatment['pdf']}")
print()

steps = [
    'Пациент открывает назначение в приложении',
    'Пациент видит окно сдачи анализов и не звонит администратору',
    'Лаборатория возвращает результат в D2',
    'Пациент смотрит результат в приложении',
    'Пациент скачивает PDF назначения из D3',
    'Пациент запрашивает статус лечения',
]
print('Шаги, которые закрывает MVP:')
for i, step in enumerate(steps, start=1):
    print(f'  {i}. {step}')
print()
print('Звонок администратору в этом сценарии не нужен: время сдачи и результат есть в приложении.')
```

Установка сервера, базы данных и библиотек Python не требуется.


## Планы по доработке

MVP закрывает просмотр назначений, результатов и окна сдачи анализов. Дальше имеет смысл развивать продукт так:

- перейти от прототипа к рабочей версии приложения с реальным backend и интеграцией с лабораторией;
- добавить онлайн-запись на сдачу анализов, а не только просмотр доступного времени;
- отправлять пациенту уведомление, когда результат анализа готов, чтобы разгрузить администратора;
- добавить напоминания о приёме лекарств и контрольном визите;
- провести юзабилити-тесты прототипа с пациентами и администратором и уточнить экраны по обратной связи;
- расширить семейный доступ: родитель видит назначения и результаты ребёнка;
- согласовать статусы назначения (создано / выполняется / завершено) между картой врача, приложением пациента и хранилищем D3;
- после реализации провести приёмочные сценарии по логической DFD на рабочем контуре, а не только на прототипе.


## Вывод по проекту

Пациенту «Виты» в приложении не хватало назначений, результатов анализов и понятного времени сдачи. Из‑за этого обращения оседали на администраторе, а бумажные направления терялись. Запись на приём и карты врача уже были; новый контур к ним не был подключён.

По интервью собраны болевые точки, на карте историй проведена граница MVP, прототип в Figma показывает, как пациент проходит новые сценарии, контекстная и логическая DFD фиксируют потоки между пациентом, врачом, лабораторией и хранилищами. Проект можно передавать в разработку как согласованную постановку первой доработки: от потребности стейкхолдеров до проверяемого интерфейса и модели процессов. Следующий шаг — реализация MVP и проверка сценариев обследования и лечения уже на рабочем приложении клиники.
